Deliverable Three: Comparing Starbucks and Dunkin' Donuts Menu for a Competitive Analysis.

```python
import pandas as pd

sbux4 = pd.read_csv("df4.csv")
dnkn4 = pd.read_csv("dnkn_clean.csv")

print(sbux4.shape)
print(dnkn4.shape)
print(sbux4.columns.tolist())
```

```python
import matplotlib.pyplot as plt

portfolio = pd.DataFrame({
    "Company": ["Starbucks", "Dunkin' Donuts"],
    "Beverage_combinations": [len(sbux4), len(dnkn4)]
})

starbucks_green = "#00704A"
dunkin_pink = "#FF5E9E"

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.barh(
    portfolio["Company"],
    portfolio["Beverage_combinations"],
    color=[starbucks_green, dunkin_pink]
)

ax.bar_label(bars, padding=3)

ax.set_title("Beverage Combination Portfolio Size")
ax.set_xlabel("Number of Beverage Combinations")
ax.set_ylabel("Company")

plt.tight_layout()
plt.show()
```

The pros and cons of the beverage combinations Starbucks has a greater variety of beverage combinations than 
Dunkin’ Donuts. It offers customers more choices, and could help Starbucks appeal to customers with different 
tastes. But then again, the more combinations you have, the higher your inventory costs and the more complex 
your operations.

```python
avg_calories = pd.DataFrame({
    "Company": ["Starbucks", "Dunkin' Donuts"],
    "Average_Calories": [
        sbux4["Calories"].mean(),
        dnkn4["Calories"].mean()
    ]
})

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(
    avg_calories["Company"],
    avg_calories["Average_Calories"],
    color=[starbucks_green, dunkin_pink]
)

ax.bar_label(bars, fmt="%.1f", padding=3)

ax.set_title("Average Beverage Calories")
ax.set_xlabel("Company")
ax.set_ylabel("Calories")

plt.tight_layout()
plt.show()
```

On average, the analysis finds Dunkin’ Donuts drinks have slightly more calories than Starbucks drinks. Starbucks 
has an average of 230.8 calories, Dunkin Donuts 253.3 calories. This suggests that Starbucks offers lower-calorie 
beverage combinations on average, although individual drink choices may vary.

```python
avg_sugar = pd.DataFrame({
    "Company": ["Starbucks", "Dunkin' Donuts"],
    "Average_Sugar": [
        sbux4["Sugar_g"].mean(),
        dnkn4["Sugars (g)"].mean()
    ]
})

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(
    avg_sugar["Company"],
    avg_sugar["Average_Sugar"],
    color=[starbucks_green, dunkin_pink]
)

ax.bar_label(bars, fmt="%.1f", padding=3)

ax.set_title("Average Beverage Sugar Content")
ax.set_xlabel("Company")
ax.set_ylabel("Sugar (g)")

plt.tight_layout()
plt.show()
```

The analysis finds that Dunkin’ Donuts drinks have a higher average sugar content than Starbucks drinks. 
Dunkin’ Donuts’ average is 42.8 grams. Starbucks is 35.0 grams on average. This difference may be important 
to customers who are trying to reduce their sugar intake.

```python
protein = pd.DataFrame({
    "Company": ["Starbucks", "Dunkin' Donuts"],
    "Average_Protein": [
        sbux4["Protein_g"].mean(),
        dnkn4["Protein (g)"].mean()
    ]
})

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(
    protein["Company"],
    protein["Average_Protein"],
    color=[starbucks_green, dunkin_pink]
)

ax.bar_label(bars, fmt="%.1f", padding=3)

ax.set_title("Average Beverage Protein Content")
ax.set_xlabel("Company")
ax.set_ylabel("Protein (g)")

plt.tight_layout()
plt.show()
```

```python
sbux_healthy = len(
    sbux4[
        (sbux4["Calories"] < 100) &
        (sbux4["Sugar_g"] < 15)
    ]
)

# Count healthier Dunkin' Donuts beverages
dnkn_healthy = len(
    dnkn4[
        (dnkn4["Calories"] < 100) &
        (dnkn4["Sugars (g)"] < 15)
    ]
)

# Create a comparison DataFrame
healthy = pd.DataFrame({
    "Company": ["Starbucks", "Dunkin' Donuts"],
    "Healthy_Beverages": [sbux_healthy, dnkn_healthy]
})

# Create bar chart
fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(
    healthy["Company"],
    healthy["Healthy_Beverages"],
    color=[starbucks_green, dunkin_pink]
)

ax.bar_label(bars, padding=3)

ax.set_title("Healthier Beverage Portfolio")
ax.set_xlabel("Company")
ax.set_ylabel("Number of Healthier Beverages")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 7))

# Starbucks beverages
plt.scatter(
    sbux4["Sugar_g"],
    sbux4["Calories"],
    color=starbucks_green,
    alpha=0.35,
    label="Starbucks"
)

# Dunkin' Donuts beverages
plt.scatter(
    dnkn4["Sugars (g)"],
    dnkn4["Calories"],
    color=dunkin_pink,
    alpha=0.35,
    label="Dunkin' Donuts"
)

plt.title("Competitive Beverage Positioning Map")
plt.xlabel("Sugar (g)")
plt.ylabel("Calories")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
top_caffeine_sbux = sbux4.sort_values(
    by="Caffeine_mg",
    ascending=False
).head(5).copy()

# Combine beverage name and size
top_caffeine_sbux["Label"] = (
    top_caffeine_sbux["Product_Name"].astype(str)
    + " ("
    + top_caffeine_sbux["Size"].astype(str)
    + ")"
)

# Create horizontal bar chart
fig, ax = plt.subplots(figsize=(12, 6))

bars = ax.barh(
    top_caffeine_sbux["Label"],
    top_caffeine_sbux["Caffeine_mg"],
    color="#00704A"
)

ax.bar_label(bars, padding=3)

ax.set_title("Top 5 Most Caffeinated Starbucks Beverages")
ax.set_xlabel("Caffeine (mg)")
ax.set_ylabel("Beverage and Size")

ax.invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
top_sugar_dnkn = dnkn4.sort_values(
    by="Sugars (g)",
    ascending=False
).head(5).copy()

# Create horizontal bar chart
fig, ax = plt.subplots(figsize=(12, 6))

bars = ax.barh(
    top_sugar_dnkn["Item"],
    top_sugar_dnkn["Sugars (g)"],
    color=dunkin_pink
)

# Display sugar values
ax.bar_label(bars, padding=3)

# Add title and labels
ax.set_title("Top 5 Most Sugary Dunkin' Donuts Beverages")
ax.set_xlabel("Sugar (g)")
ax.set_ylabel("Beverage")

# Show highest sugar content at the top
ax.invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
avg_sodium = pd.DataFrame({
    "Company": ["Starbucks", "Dunkin' Donuts"],
    "Average_Sodium": [
        sbux4["Sodium_mg"].mean(),
        dnkn4["Sodium (mg)"].mean()
    ]
})

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(
    avg_sodium["Company"],
    avg_sodium["Average_Sodium"],
    color=[starbucks_green, dunkin_pink]
)

ax.bar_label(bars, fmt="%.1f", padding=3)

ax.set_title("Average Sodium Content of Starbucks and Dunkin' Donuts Beverages")
ax.set_xlabel("Company")
ax.set_ylabel("Average Sodium (mg)")

plt.tight_layout()
plt.show()
```

Another huge factor when analyzing nutritional information is sodium so I thought I would compare the average 
sodium content of Starbucks vs Dunkin’ Donuts beverages. The results indicate that Starbucks drinks contain an 
average of 142.8 mg of sodium, whereas Dunkin’ Donuts drinks contain an average of 139.4 mg of sodium. Starbucks 
is a little higher on the average but it is not a big difference. This comparison helps provide a better 
understanding of the nutritional differences between the two companies beyond calories, sugar, and protein.
